## Kreiranje objekta list s 100 naključnimi vrednostmi

Naredili smo list s 100 naključnimi števili s ***for*** zanko

```python
# create a list of 100 random numbers between 1 and 1000
random_numbers=[]
for stevec in range(100):
  print("Korak "+str(stevec))
  vrednost=rd.randint(1,1000)
  print(vrednost)
  random_numbers.append(vrednost)

random_numbers[:10]
```

##Prikaz f-stringa

```python
vrednost=200
# print("Na predstavi se je zbralo"+ str(vrednost)+" ljudi.")
print(f"Na predstavi se je zbralo {vrednost} ljudi.")
```

## Ali je random funkcija res random

```python
stevecSodoStevilo=0
stevecLihoStevilo=0
for stevilka in random_numbers:
  if stevilka%2==0:
    stevecSodoStevilo+=1
  else:
    stevecLihoStevilo+=1

print(f"odstotek sodih števil je {stevecSodoStevilo/len(random_numbers)*100}, lihih pa {100-stevecSodoStevilo/len(random_numbers)*100}!")
```

## Primer diabetesa

```python
df1=pd.read_csv("https://raw.githubusercontent.com/urosgodnov/datasets/refs/heads/master/diabetes.csv")
df1
```

```python
df1.describe()
```

```python
# replace 0 with NaN for all variables except Class_variable
import numpy as np

columns=df1.columns
for column in columns:
  if column not in ["Class_variable","Number_of_times_pregnant"]:
    df1[column]=df1[column].replace(0,np.nan)

df1.describe()
```

```python
df1
```

```python
# @title Plasma_glucose_concentration vs Diastolic_blood_pressure

from matplotlib import pyplot as plt
df1.plot(kind='scatter', x='Plasma_glucose_concentration', y='Diastolic_blood_pressure', s=32, alpha=.8)
plt.gca().spines[['top', 'right',]].set_visible(False)
```
